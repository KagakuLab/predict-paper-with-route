### 1. Train model

```python
data_train = pd.read_csv("../oprd/oprd_data/OPRDpaper_1997_2022.csv")
text_train = [t + a for t, a in zip(data_train["title"], data_train["abstract"])]
label_train = [i for i in data_train["has_route"]]
```

```python
model = Pipeline([("tfidf", TfidfVectorizer()), ("lr", LogisticRegression(max_iter=1000, random_state=42))])
model.fit(text_train, label_train)
```

### 2. Predict new papers

```python
data_new_1 = pd.read_csv("journal_data/openalex_chem_metadata_1.csv")
data_new_2 = pd.read_csv("journal_data/openalex_chem_metadata_2.csv")
```

```python
data_new = pd.concat([data_new_1, data_new_2], axis=0)
```

```python
text_new = [str(t) + str(a) for t, a in zip(data_new["title"], data_new["abstract"])]
```

```python
y_pred = list(model.predict(text_new))
y_prob = list(model.predict_proba(text_new)[:, 1])
```

```python
tot_papers, route_papers = len(y_pred), sum(y_pred).item()
```

```python
print(f"Total papers: {tot_papers}")
print(f"Paper with route: {route_papers} ({100*route_papers/tot_papers:.1f}%)")
```

```python
df_pred = pd.DataFrame()
df_pred["text"], df_pred["doi"], df_pred["prob"] = text_new, data_new["doi"], y_prob
df_pred = df_pred.sort_values(by="prob", ascending=False)
```
